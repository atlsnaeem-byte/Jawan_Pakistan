### Assignment 10 - Filtering & Selecting Data with Pandas

```python
# Importing the pandas library
import pandas as pd

# Reading the excel file into a DataFrame
df = pd.read_excel("retail_supermarket.xlsx")

# Looking at the first 5 rows
df.head()
```

```python
# Checking shape and column names of the data
print("Rows and Columns:", df.shape)
print()
print(df.columns)
```

#### 1- Find record if Sales greater than 500 and Quantity greater than 5?


```python
# Using & for AND condition (both conditions must be true)
q1 = df[(df["Sales"] > 500) & (df["Quantity"] > 5)]

print("Total records found:", len(q1))
q1.head()
```

#### 2- Find records if Discount is more than 0.5 (50%) or Profit is negative.


```python
# Using | for OR condition (any one condition can be true)
q2 = df[(df["Discount"] > 0.5) | (df["Profit"] < 0)]

print("Total records found:", len(q2))
q2.head()
```

#### 3- Display Ship Mode and Profit columns if region is East using loc

```python
# loc is used as df.loc[row_condition, column_names]
q3 = df.loc[df["Region"] == "East", ["Ship Mode", "Profit"]]

print("Total records found:", len(q3))
q3.head()
```

#### 4- How many records belong to the "Corporate" segment and the "Central" region?


```python
# First filtering both conditions, then counting the rows
q4 = df[(df["Segment"] == "Corporate") & (df["Region"] == "Central")]

print("Number of records:", len(q4))
q4.head()
```

#### 5- Find all orders in the "Furniture" category but not in "Chairs" sub-category.


```python
# != means "not equal to"
q5 = df[(df["Category"] == "Furniture") & (df["Sub-Category"] != "Chairs")]

print("Total records found:", len(q5))
q5.head()
```

#### 6- Which orders have discount (>0.3) and still earned profit (>0)?


```python
# High discount but still in profit
q6 = df[(df["Discount"] > 0.3) & (df["Profit"] > 0)]

print("Total records found:", len(q6))
q6.head()
```

#### 7- Find high-profit orders where Quantity > 5 and Category == 'Technology'.


```python
# Taking profit greater than 0 as "high-profit" orders
q7 = df[(df["Quantity"] > 5) & (df["Category"] == "Technology") & (df["Profit"] > 0)]

print("Total records found:", len(q7))
q7.head()
```

#### 8- Select rows 0-4 and columns 'Ship Mode', 'Segment', 'Sales' by position using iloc


```python
# iloc works with index numbers only
# Ship Mode = column 0, Segment = column 1, Sales = column 9
print(list(df.columns))

q8 = df.iloc[0:5, [0, 1, 9]]
q8
```

#### 9- Show Home Office segment orders with Sales > 1000 and Quantity > 3.


```python
# Three conditions joined with &
q9 = df[(df["Segment"] == "Home Office") & (df["Sales"] > 1000) & (df["Quantity"] > 3)]

print("Total records found:", len(q9))
q9.head()
```

#### 10- Find orders where Profit Margin > 0.3
##### Profit Margin = (Profit/Sales)

```python
# Making a new column for profit margin
df["Profit Margin"] = df["Profit"] / df["Sales"]

# Now filtering on that new column
q10 = df[df["Profit Margin"] > 0.3]

print("Total records found:", len(q10))
q10.head()
```
