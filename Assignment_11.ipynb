# Assignment 11 - Cars Dataset

#### Import libraries and load the data

```python
import pandas as pd

df = pd.read_excel("cars.xlsx")
df.head()
```

```python
# how many rows and columns do we have?
df.shape
```

```python
# column names and data types
df.info()
```

#### Check Duplicate Records

```python
# True means the row is a copy of an earlier row
df.duplicated().sum()
```

#### Drop Duplicate Records

```python
df = df.drop_duplicates()

# check again, it should be 0 now
print(df.duplicated().sum())
print(df.shape)
```

#### Check missing or null values

```python
df.isnull().sum()
```

#### Drop missing or null values

```python
df = df.dropna()

# check again, all columns should be 0
print(df.isnull().sum())
print(df.shape)
```

#### Show top 10 average MSRP (Price) per car brand (rounded figure)?

```python
top10_price = df.groupby("Make")["MSRP"].mean().round()
top10_price = top10_price.sort_values(ascending=False).head(10)
top10_price
```

#### Show the average MSRP (price) per years (rounded figure)?

```python
price_per_year = df.groupby("Year")["MSRP"].mean().round()
price_per_year
```

#### Show average Engine HP (power) rounded figure and MPG (efficiency) rounded figure per year?

Hint: Average of (city mpg + highway MPG)

```python
# first make a new column for average MPG
df["Average MPG"] = (df["city mpg"] + df["highway MPG"]) / 2

hp_mpg = df.groupby("Year")[["Engine HP", "Average MPG"]].mean().round()
hp_mpg
```

#### What percentage of cars are Compact, Midsize, or Large?

```python
size_percent = df["Vehicle Size"].value_counts(normalize=True) * 100
size_percent = size_percent.round(2)
size_percent
```

#### Which vehicle styles are the most sold (most common) in the dataset?

```python
style_count = df["Vehicle Style"].value_counts()
style_count
```

```python
# the most common one
style_count.head(1)
```

#### Which car brands have the most vehicles priced at $40,000 or more and with a popularity score of at least 1000?

```python
expensive = df[(df["MSRP"] >= 40000) & (df["Popularity"] >= 1000)]

brand_count = expensive["Make"].value_counts()
brand_count.head(10)
```
