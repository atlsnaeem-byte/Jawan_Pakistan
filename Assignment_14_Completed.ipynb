## Assignment 14 — SQL with SQLite

```python
import sqlite3
import pandas as pd
import os

# Use database.sqlite from the current folder.
# The fallback path allows the notebook to run in this environment as well.
DB_PATH = "database.sqlite"
if not os.path.exists(DB_PATH):
    DB_PATH = "/mnt/data/database.sqlite"

conn = sqlite3.connect(DB_PATH)

def sql_query(query):
    return pd.read_sql_query(query, conn)
```

#### **1. What is the average TotalPayBenefits for each JobTitle? exclude ('Not Provided', 'Not provided')**

```python
query_1 = '''
SELECT
    JobTitle,
    ROUND(AVG(TotalPayBenefits), 2) AS Avg_TotalPayBenefits
FROM Salaries
WHERE JobTitle NOT IN ('Not Provided', 'Not provided')
GROUP BY JobTitle
ORDER BY Avg_TotalPayBenefits DESC;
'''
sql_query(query_1)
```

#### **2. Show The JobTitle and Overtime Pay for All Employees With Overtime Pay greater than 50000**

```python
query_2 = '''
SELECT
    EmployeeName,
    JobTitle,
    CAST(OvertimePay AS REAL) AS OvertimePay
FROM Salaries
WHERE CAST(OvertimePay AS REAL) > 50000
ORDER BY OvertimePay DESC;
'''
sql_query(query_2)
```

#### **3. What is the total number of unique employees? exclude ('Not Provided', 'Not provided')**

```python
query_3 = '''
SELECT
    COUNT(DISTINCT EmployeeName) AS Total_Unique_Employees
FROM Salaries
WHERE EmployeeName NOT IN ('Not Provided', 'Not provided');
'''
sql_query(query_3)
```

#### **4.Show The Top 10 highest Paid Employees**

```python
query_4 = '''
SELECT
    EmployeeName,
    JobTitle,
    TotalPayBenefits
FROM Salaries
WHERE EmployeeName NOT IN ('Not Provided', 'Not provided')
ORDER BY TotalPayBenefits DESC
LIMIT 10;
'''
sql_query(query_4)
```

#### **5.Show All Employees Ordered By Their TotalPay Benefits In Descending Order**

```python
query_5 = '''
SELECT
    EmployeeName,
    JobTitle,
    TotalPayBenefits
FROM Salaries
WHERE EmployeeName NOT IN ('Not Provided', 'Not provided')
ORDER BY TotalPayBenefits DESC;
'''
sql_query(query_5)
```
