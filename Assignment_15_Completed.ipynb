# Assignment 15 — SQL JOIN Practice

```python
import sqlite3
import pandas as pd
from pathlib import Path

# Use the database from the same folder as this notebook.
db_path = Path("factbook(1).db")
if not db_path.exists():
    db_path = Path("/mnt/data/factbook(1).db")

conn = sqlite3.connect(db_path)

def sql_query(query):
    return pd.read_sql_query(query, conn)
```

1. List all cities where the area is greater than 5000, along with their population.

```python
# 1. Cities whose corresponding country's area is greater than 5000,
#    along with each city's population.
q1 = '''
SELECT
    c.name AS city_name,
    c.population AS city_population
FROM cities AS c
JOIN facts AS f
    ON c.facts_id = f.id
WHERE f.area > 5000
ORDER BY c.population DESC;
'''

sql_query(q1)
```

2. List the names of capital cities where the death rate is greater than 2% and the
birth rate is less than 20%

```python
# 2. Capital cities where death rate > 2% and birth rate < 20%.
q2 = '''
SELECT
    c.name AS capital_city,
    f.death_rate,
    f.birth_rate
FROM cities AS c
JOIN facts AS f
    ON c.facts_id = f.id
WHERE c.capital = 1
  AND f.death_rate > 2
  AND f.birth_rate < 20
ORDER BY c.name;
'''

sql_query(q2)
```

3. List the cities with a population growth rate between 1% and 5%, sorted in desc
by population growth.

```python
# 3. Cities with population growth between 1% and 5%,
#    sorted by population growth in descending order.
q3 = '''
SELECT
    c.name AS city_name,
    f.population_growth
FROM cities AS c
JOIN facts AS f
    ON c.facts_id = f.id
WHERE f.population_growth BETWEEN 1 AND 5
ORDER BY f.population_growth DESC;
'''

sql_query(q3)
```

4. Find countries that start with &quot;A&quot; and do not have any cities listed in the cities
table. The query should return the country name and city name. Using left join

```python
# 4. Countries starting with "A" that have no cities listed.
#    LEFT JOIN is used as requested.
q4 = '''
SELECT
    f.name AS country_name,
    c.name AS city_name
FROM facts AS f
LEFT JOIN cities AS c
    ON f.id = c.facts_id
WHERE f.name LIKE 'A%'
  AND c.id IS NULL
ORDER BY f.name;
'''

sql_query(q4)
```

5. List all capital cities and their corresponding death rate using a LEFT JOIN.

```python
# 5. All capital cities and their corresponding death rate using LEFT JOIN.
q5 = '''
SELECT
    c.name AS capital_city,
    f.death_rate
FROM cities AS c
LEFT JOIN facts AS f
    ON c.facts_id = f.id
WHERE c.capital = 1
ORDER BY c.name;
'''

sql_query(q5)
```
