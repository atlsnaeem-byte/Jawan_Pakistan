Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

### Answer 1 — Using a Class (OOP)

**Step 1 — Defining the `Employee` class**

```python
class Employee:
    """A class to store and manage the information of a single employee."""

    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):
        """Store all the details of the employee."""
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        """Print the complete information about the employee."""
        print("-" * 45)
        print("EMPLOYEE DETAILS")
        print("-" * 45)
        print(f"Name              : {self.name}")
        print(f"Contact Number    : {self.number}")
        print(f"Address           : {self.address}")
        print(f"Post              : {self.post}")
        print(f"Salary            : Rs. {self.salary:,.0f}")
        print(f"Year of Joining   : {self.year_of_joining}")
        print(f"Education         : {self.education}")
        print(f"Experience        : {self.year_of_experience} years")
        print("-" * 45)

    def update_year_of_experience(self, years=1):
        """Add the given number of years to the employee's experience."""
        if years < 0:
            print("Experience cannot be decreased.")
            return
        old_experience = self.year_of_experience
        self.year_of_experience += years
        print(f"{self.name}: experience updated from {old_experience} "
              f"to {self.year_of_experience} years.")

    def increase_salary(self, percent=10):
        """Increase the employee's salary by the given percentage."""
        if percent < 0:
            print("Increment percentage cannot be negative.")
            return
        old_salary = self.salary
        increment = self.salary * percent / 100
        self.salary += increment
        print(f"{self.name}: salary increased by {percent}% "
              f"(Rs. {increment:,.0f}) -> Rs. {old_salary:,.0f} "
              f"to Rs. {self.salary:,.0f}.")
```

**Step 2 — Creating `employee101` and printing four attributes individually**

```python
# Creating an instance of the Employee class
employee101 = Employee(
    name="Ali Raza",
    number="0300-1234567",
    address="House 12, Block B, Gulshan-e-Iqbal, Karachi",
    salary=85000,
    post="Software Engineer",
    year_of_joining=2021,
    education="BS Computer Science",
    year_of_experience=3
)

# Printing four attributes individually
print("Name      :", employee101.name)
print("Post      :", employee101.post)
print("Salary    :", employee101.salary)
print("Education :", employee101.education)
```

**Step 3 — Calling all the methods**

```python
# Calling all the methods of employee101

employee101.describe_employee()

print()
employee101.increase_salary(15)          # 15% increment
employee101.update_year_of_experience(2) # 2 more years of experience

print("\nAfter the updates:")
employee101.describe_employee()
```

### Answer 2 — Same Task Without Classes / OOP

Here every employee is stored in a **dictionary** instead of an object, and the
methods are replaced by ordinary **functions** that take the dictionary as an argument.

**Step 1 — Defining the functions**

```python
def make_employee(name, number, address, salary, post,
                  year_of_joining, education, year_of_experience):
    """Return a dictionary holding all details of one employee."""
    return {
        "name": name,
        "number": number,
        "address": address,
        "salary": salary,
        "post": post,
        "year_of_joining": year_of_joining,
        "education": education,
        "year_of_experience": year_of_experience,
    }


def describe_employee(employee):
    """Print the complete information about the employee."""
    print("-" * 45)
    print("EMPLOYEE DETAILS")
    print("-" * 45)
    print(f"Name              : {employee['name']}")
    print(f"Contact Number    : {employee['number']}")
    print(f"Address           : {employee['address']}")
    print(f"Post              : {employee['post']}")
    print(f"Salary            : Rs. {employee['salary']:,.0f}")
    print(f"Year of Joining   : {employee['year_of_joining']}")
    print(f"Education         : {employee['education']}")
    print(f"Experience        : {employee['year_of_experience']} years")
    print("-" * 45)


def update_year_of_experience(employee, years=1):
    """Add the given number of years to the employee's experience."""
    if years < 0:
        print("Experience cannot be decreased.")
        return
    old_experience = employee["year_of_experience"]
    employee["year_of_experience"] = old_experience + years
    print(f"{employee['name']}: experience updated from {old_experience} "
          f"to {employee['year_of_experience']} years.")


def increase_salary(employee, percent=10):
    """Increase the employee's salary by the given percentage."""
    if percent < 0:
        print("Increment percentage cannot be negative.")
        return
    old_salary = employee["salary"]
    increment = old_salary * percent / 100
    employee["salary"] = old_salary + increment
    print(f"{employee['name']}: salary increased by {percent}% "
          f"(Rs. {increment:,.0f}) -> Rs. {old_salary:,.0f} "
          f"to Rs. {employee['salary']:,.0f}.")
```

**Step 2 — Creating `employee101` and printing four attributes individually**

```python
# Creating one employee record
employee101 = make_employee(
    name="Ali Raza",
    number="0300-1234567",
    address="House 12, Block B, Gulshan-e-Iqbal, Karachi",
    salary=85000,
    post="Software Engineer",
    year_of_joining=2021,
    education="BS Computer Science",
    year_of_experience=3
)

# Printing four attributes individually
print("Name      :", employee101["name"])
print("Post      :", employee101["post"])
print("Salary    :", employee101["salary"])
print("Education :", employee101["education"])
```

**Step 3 — Calling all the functions**

```python
# Calling all the functions for employee101

describe_employee(employee101)

print()
increase_salary(employee101, 15)
update_year_of_experience(employee101, 2)

print("\nAfter the updates:")
describe_employee(employee101)
```
