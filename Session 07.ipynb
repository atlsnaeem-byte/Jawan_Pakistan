## ***Why do we use for loop in python?***

***A for loop in Python is used to iterate over a sequence (like a list, tuple, string, or range) and execute a block of code repeatedly for each item in the sequence. It helps automate repetitive tasks, making the code more concise and efficient***

```python
# Tuple
my_tuple = (1, 2, 3, 4, 5)
my_tuple
```

```python
print(my_tuple[0])
print(my_tuple[1])
print(my_tuple[2])
print(my_tuple[3])
print(my_tuple[4])
```

```python
for x in my_tuple:
    print(x)
```

```python
for name in my_tuple:
    print(name)
```

```python
# List
my_list = ['apple', 'banana', 'cherry']
my_list
```

```python
for fruit in my_list:
    print(fruit)
```

```python
for index, city in enumerate(cities):
    print(index, city)
```

## ***Nested for loop***

```python
cities = ['karachi','hyderabad', 'islamabad', 'quetta','lahore']
visited_cities = ['islamabad', 'multan','faisalabad','quetta','peshawar']
```

```python
for city in cities:
    for visit in visited_cities:
        print(city, visit)
    print(' ')
```

```python
print(cities)
print(visited_cities)
```

```python
for city in cities:
    print(city)
    for visit in visited_cities:
        print(visit)
    print(' ')
```

```python
week_days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
time_slots = ['9AM-10AM', '10AM-11AM', '11AM-12PM']
```

```python
for day in week_days:
    print(day)
    for slot in time_slots:
        print(slot)
    print('------')
```

```python
for day in week_days:
    for slot in time_slots:
        print(day,slot)
    print('------')
```

```python
cities = []
count = ''

while count != 'quit':
    city = input('enter a city')
    if city == 'quit':
        count = 'quit'
    else:
        cities.append(city)
cities 
```
